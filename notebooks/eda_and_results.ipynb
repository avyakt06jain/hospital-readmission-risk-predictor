## SQL cohort analysis

```python
for name in ["cohort_by_diagnosis", "cohort_by_payer", "cohort_by_los", "cohort_by_inpatient"]:
    print(name)
    display(pd.read_csv(ROOT / "data/processed" / f"{name}.csv").head(10))
```

```python
metrics = json.loads((ROOT / "plots/metrics.json").read_text())
print(json.dumps({k: v for k, v in metrics.items() if k != "xgboost" and k != "lightgbm"}, indent=2))
print("XGBoost AUROC", metrics["xgboost"]["auroc"])
print("LightGBM AUROC", metrics["lightgbm"]["auroc"])
for p in ["cohort_readmission_rates.png", "roc_curve.png", "pr_curve.png", "confusion_matrix.png", "shap_beeswarm.png", "shap_waterfall_example.png"]:
    display(Image(filename=str(ROOT / "plots" / p)))
```
